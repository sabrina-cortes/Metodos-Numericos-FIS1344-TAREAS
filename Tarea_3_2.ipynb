Repita el **problema de la urna pero con reemplazo**. Es decir, luego de sacar cada bola esta se vuelve a poner en la urna. Si se extraen cuatro bolas de la urna, ¿cuál es la probabilidad de que todas sean rojas? Responda este problema escribiendo un código, **modificando las funciones escritas en la clase para tomar en cuenta el reemplazo.**

En este problema usamos los mismos datos del enunciado 3.1 que son parte del problema realizado por el profesor, es decir, **23 bolas en total: 8 blancas, 6 azules y 9 rojas.**

La diferencia con el problema anterior es que ahora las extracciones se realizan **con reemplazo**. Esto significa que, después de sacar una bola, esta vuelve a colocarse en la urna antes de realizar la siguiente extracción.

Por lo tanto, la **cantidad total de bolas se mantiene constante** en cada extracción y una misma bola puede aparecer nuevamente en una extracción posterior.

En el caso sin reemplazo se utilizaban permutaciones, ya que una bola que había sido extraída no podía volver a aparecer. En cambio, al existir reemplazo, debemos permitir repeticiones dentro de las secuencias de resultados.

Por esta razón utilizamos:

```python
itertools.product(urna, repeat=4)
```

en lugar de:

```python
itertools.permutations(urna, 4)
```

De esta forma se generan todas las secuencias posibles de cuatro extracciones en las que una misma bola puede aparecer más de una vez.

```python
import itertools

# Función que usó el profesor para asociar
# cada color con los números de las bolitas
def unir(color, numeros):
    return {color + numero for numero in numeros}

# Construimos la urna:
# 8 blancas, 6 azules y 9 rojas
urna = unir('B', '12345678') | unir('A', '123456') | unir('R', '123456789')

# Como ahora hay reemplazo, una misma bola puede
# volver a aparecer en una extracción posterior.
# product(..., repeat=4) genera todas las secuencias
# posibles de cuatro extracciones con reemplazo.
espacio = set(itertools.product(urna, repeat=4))

# Función que revisa si las cuatro bolas extraídas son rojas
def todos_rojos(evento):

    # Tomamos la primera letra de cada bola
    # (que indica su color)
    colores = [i[0] for i in evento]

    # El evento es favorable si aparecen cuatro rojas
    return colores.count('R') == 4

# Seleccionamos del espacio todos los eventos
# donde las cuatro bolas son rojas
rojos = {e for e in espacio if todos_rojos(e)}

# Función de probabilidad usada en el
# problema del profesor
def prob(evento, espacio):
    return len(evento & espacio) / len(espacio)

# Calculamos la probabilidad
probabilidad = prob(rojos, espacio)

print(probabilidad)
```

Como existen 9 bolas rojas entre un total de 23, la probabilidad de obtener una bola roja es:

$$
P(R)=\frac{9}{23}
$$

Como hay reemplazo, la probabilidad se mantiene igual en cada extracción:

$$
P(R,R,R,R)
=
\frac{9}{23}
\cdot
\frac{9}{23}
\cdot
\frac{9}{23}
\cdot
\frac{9}{23}
$$

Por lo tanto:

$$
P(R,R,R,R)=\left(\frac{9}{23}\right)^4
$$

y finalmente:

$$
P \approx 0.0234
$$
