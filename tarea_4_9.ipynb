Idea central de la tarea:

En esta tarea se compara la probabilidad obtenida mediante una simulación **Monte Carlo** con la aproximación dada por el **Teorema Central del Límite**.

Se consideran 20 números aleatorios independientes extraídos de una distribución uniforme entre 0 y 1.

Definimos la suma:

$$
X = x_1 + x_2 + \cdots + x_{20}
$$

Se desea estimar las probabilidades:

$$
P(9<X<10)
$$

y

$$
P(15<X<16)
$$

Primero realizamos una simulación mediante **Monte Carlo.**

```python
# Importamos NumPy para generar números aleatorios
# y realizar operaciones numéricas.
import numpy as np

# Importamos erf desde scipy.special,
# ya que se utilizará para calcular probabilidades
# de la distribución gaussiana.
from scipy.special import erf
```

Para realizar la simulación se generan muchas realizaciones.

En cada realización:

1. Se generan 20 números aleatorios uniformemente distribuidos entre 0 y 1.
2. Se suman los 20 valores.
3. Se verifica si la suma pertenece al intervalo $(9,10)$ o al intervalo $(15,16)$.

La probabilidad se estima mediante:

$$
P(A) \approx \frac{N_A}{N}
$$

donde $N$ es el número total de simulaciones y $N_A$ es el número de veces que ocurre el evento de interés.

```python
# Número total de simulaciones,
# más simulaciones es mejor
N = 1000000

# Contadores de eventos favorables
contador_9_10 = 0
contador_15_16 = 0

# Repetimos el experimento N veces
for i in range(N):

    # Generamos 20 números aleatorios uniformes
    # en el intervalo [0,1)
    numeros = np.random.random(20)
    suma = np.sum(numeros)

    # Contamos si la suma está entre 9 y 10
    if 9 < suma < 10:
        contador_9_10 += 1

    # Contamos si la suma está entre 15 y 16
    if 15 < suma < 16:
        contador_15_16 += 1


# Estimamos las probabilidades mediante frecuencia relativa
p_mc_9_10 = contador_9_10 / N
p_mc_15_16 = contador_15_16 / N

print("Probabilidad Monte Carlo para 9 < X < 10:", p_mc_9_10)
print("Probabilidad Monte Carlo para 15 < X < 16:", p_mc_15_16)
```

Ahora realizamos la aproximación mediante el método del Teorema Central del Límite:

Cada variable $x_i$ sigue una distribución uniforme entre 0 y 1.

Para una distribución uniforme $U(0,1)$:

$$
\mu = \frac{1}{2}
$$

y

$$
\sigma^2 = \frac{1}{12}
$$

Como:

$$
X = \sum_{i=1}^{20} x_i
$$

la media de la suma es:

$$
\mu_X = 20\mu
$$

por lo tanto:

$$
\mu_X = 20\left(\frac{1}{2}\right)=10
$$

La varianza de la suma de variables independientes es la suma de sus varianzas:

$$
\sigma_X^2 = 20\sigma^2
$$

entonces:

$$
\sigma_X^2 = 20\left(\frac{1}{12}\right)
           = \frac{5}{3}
$$

y la desviación estándar es:

$$
\sigma_X = \sqrt{\frac{5}{3}}
$$

Por el Teorema Central del Límite, la distribución de la suma puede aproximarse mediante una distribución gaussiana:

$$
X \approx N\left(10,\frac{5}{3}\right)
$$

```python
import numpy as np

# Parámetros de cada variable uniforme y de la suma
mu_individual = 1/2
var_individual = 1/12
n = 20

mu = n * mu_individual
varianza = n * var_individual
sigma = np.sqrt(varianza)

print("Media de la suma =", mu)
print("Varianza de la suma =", varianza)
print("Desviación estándar =", sigma)
```

Para calcular la probabilidad de una distribución gaussiana entre dos valores se utiliza la función error.

La función acumulada de una gaussiana puede escribirse como:

$$
F(x)=
\frac{1}{2}
\left[
1+
\operatorname{erf}
\left(
\frac{x-\mu}{\sqrt{2}\sigma}
\right)
\right]
$$

Por lo tanto:

$$
P(a<X<b)=F(b)-F(a)
$$

```python
# Definimos la función de distribución acumulada
# para una gaussiana con media mu y desviación sigma.
def acumulada_gaussiana(x, mu, sigma):

    return 0.5 * (
        1 + erf((x - mu) / (np.sqrt(2) * sigma))
    )


# Probabilidad gaussiana entre 9 y 10
p_gauss_9_10 = (
    acumulada_gaussiana(10, mu, sigma)
    - acumulada_gaussiana(9, mu, sigma)
)

# Probabilidad gaussiana entre 15 y 16
p_gauss_15_16 = (
    acumulada_gaussiana(16, mu, sigma)
    - acumulada_gaussiana(15, mu, sigma)
)

print("Aproximación gaussiana para 9 < X < 10:", p_gauss_9_10)
print("Aproximación gaussiana para 15 < X < 16:", p_gauss_15_16)
```

Ahora estimamos el error asociado a la simulación Monte Carlo

Como cada simulación puede considerarse un experimento con dos resultados:

- el evento ocurre,
- el evento no ocurre,

la estimación de la probabilidad se comporta como una proporción binomial.

El error estadístico de una probabilidad estimada mediante Monte Carlo puede aproximarse por:

$$
\sigma_P =
\sqrt{
\frac{P(1-P)}{N}
}
$$

donde $P$ es la probabilidad estimada y $N$ es el número total de simulaciones.

```python
# Calculamos el error estadístico de Monte Carlo
# para el intervalo entre 9 y 10.
error_mc_9_10 = np.sqrt(
    p_mc_9_10 * (1 - p_mc_9_10) / N
)

# Calculamos el error estadístico de Monte Carlo
# para el intervalo entre 15 y 16.
error_mc_15_16 = np.sqrt(
    p_mc_15_16 * (1 - p_mc_15_16) / N
)

print("Error MC para 9 < X < 10:", error_mc_9_10)
print("Error MC para 15 < X < 16:", error_mc_15_16)
```

Para comparar ambos resultados calculamos el error relativo:

$$
E_r =
\frac{
|P_{\mathrm{MC}}-P_{\mathrm{Gauss}}|
}{
|P_{\mathrm{Gauss}}|
}
$$

También puede expresarse como porcentaje:

$$
E_{\%}=100E_r
$$

```python
# Error relativo para el intervalo 9 < X < 10
error_rel_9_10 = (
    abs(p_mc_9_10 - p_gauss_9_10)
    / abs(p_gauss_9_10)
)

# Error relativo para el intervalo 15 < X < 16
error_rel_15_16 = (
    abs(p_mc_15_16 - p_gauss_15_16)
    / abs(p_gauss_15_16)
)


print("Error relativo 9 < X < 10:", error_rel_9_10)
print("Error porcentual 9 < X < 10:", 100*error_rel_9_10, "%")

print()

print("Error relativo 15 < X < 16:", error_rel_15_16)
print("Error porcentual 15 < X < 16:", 100*error_rel_15_16, "%")
```

Para el intervalo $9<X<10$, los resultados de Monte Carlo y de la aproximación gaussiana son bastante cercanos. Esto tiene sentido porque el intervalo está cerca de la media de la distribución, que es:

$$
\mu_X=10
$$

En cambio, el intervalo $15<X<16$ está mucho más alejado de la media. En esta zona la probabilidad es muy pequeña, por lo que en la simulación aparecen pocos casos favorables y el error relativo aumenta.

Además, la aproximación gaussiana dada por el Teorema Central del Límite funciona mejor cerca de la zona central de la distribución que en sus extremos.

Por esto, se obtiene una mejor concordancia entre ambos métodos para $9<X<10$ que para $15<X<16$.
