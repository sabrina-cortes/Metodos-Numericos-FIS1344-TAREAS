Código 1:

```python
n = 1000
s = 0
for i in range(n):
    for j in range(n):
        if i < j:
            s = s + i + j
```

El primer **ciclo for** realiza \(n\) iteraciones. Para cada una de ellas, el segundo ciclo también realiza \(n\) iteraciones. Por lo tanto, se consideran $$n ⋅ n= n^2$$ pares (i,j).

La condición **if i < j** hace que la suma se realice solo en parte de esos casos, pero la condición igualmente debe evaluarse para los $$(n^2)$$ pares. Por lo tanto, no cambia el orden de crecimiento.

$$ T(n)\propto n^2 $$ $$ {O(n^2)} $$

Por lo tanto:

**c=2**
	​


Código 2:

```python
n = 1000
s = 0
for i in range(n):
    for j in range(i):
        s = s + i + j
```

El primer ciclo recorre:

$$ i=0,1,2,\ldots,n-1 $$

El segundo ciclo depende del valor de (i), ya que range(i) contiene (i) elementos.

Por lo tanto, el número total de iteraciones del ciclo interior es:

$$ 0+1+2+\cdots +(n-1) $$

Esta suma corresponde a:

$$ \frac{n(n-1)}{2} = \frac{n^2-n}{2} $$

Para (n) grande, el término dominante es $$(n^2)$$ Las constantes y términos de menor grado no afectan el orden:

$$ {O(n^2)} $$

entonces:

**c=2**

Código 3:

```python
n = 1000
s = 0
for i in range(n):
    s = s + i + sum(range(n))
```

El ciclo **for i in range(n)** se ejecuta n veces.
Dentro de cada iteración aparece **sum(range(n))**, tal que range(n) contiene (n) elementos y sum() debe recorrerlos para realizar la suma, por lo tanto, esta operación tiene un valor proporcional a (n).

Como esa suma se calcula nuevamente en cada una de las (n) iteraciones del ciclo $$n⋅n=n^2$$ y por lo tanto, $${O(n^2)}$$
con:
**c=2**

Código 4: "aumente n y vea cómo cambia el número de pasos s. Grafique n contra s e intente deducir a qué corresponde".

```python
import numpy as np
import matplotlib.pyplot as plt

# Distintos valores de n para probar el mismo algoritmo
valores_n = np.array([1000, 5000, 10000, 50000, 100000, 500000])

# Lista donde guardaremos el número de pasos
pasos = []

# Repetimos el código original para cada valor de n
for valor in valores_n:

    # En cada repetición n toma un valor distinto
    n = valor

    # Utilizamos el código entregado en el enunciado
    small = 0
    p = 146

    attempt = n//2
    not_found = True
    s = 0

    while not_found:

        s = s + 1

        if attempt > p:
            n = attempt

        elif attempt < p:
            small = attempt

        else:
            print("Encontré", attempt, "luego de", s,
                  "pasos para n =", valor)
            break

        attempt = (n + small)//2

    # Guardamos el número de pasos obtenido
    pasos.append(s)


# Graficamos los resultados
plt.scatter(valores_n, pasos)

plt.xlabel("n")
plt.ylabel("Número de pasos")
plt.title("Número de pasos en función de n")

plt.show()
```

En este algoritmo, el objetivo es encontrar un valor (p) dentro de un intervalo cuyo límite superior inicial es (n).

En cada iteración, el algoritmo prueba el punto medio del intervalo actual:

$$\text{attempt} = \frac{n+\text{small}}{2}$$

Dependiendo de si attempt es mayor o menor que (p), se descarta aproximadamente la mitad del intervalo de búsqueda. Este proceso de reducción del intervalo se puede observar de la siguiente manera:

$$n, \frac{n}{2}, \frac{n}{4}, \frac{n}{8}, \ldots$$

Después de \(s\) pasos, el tamaño del intervalo se reduce aproximadamente a:

$$\frac{n}{2^s}$$

El intervalo de búsqueda continúa reduciéndose aproximadamente a la mitad hasta alcanzar un tamaño del orden de 1. Por ello:

$$\frac{n}{2^s} \approx 1$$

Despejando \(n\) de la ecuación anterior, obtenemos:

$$n \approx 2^s$$

Aplicando el logaritmo en base 2 a ambos lados:

$$s \approx \log_2(n)$$

Este resultado fundamental indica que el número de pasos (s) crece de forma logarítmica con respecto a \(n\). Es decir, incluso si \(n\) aumenta considerablemente, el número de pasos aumenta de manera mucho más lenta.

El gráfico obtenido al probar distintos valores de \(n\) es consistente con un crecimiento logarítmico: al aumentar considerablemente \(n\), el número de pasos crece lentamente.

Por lo tanto, la complejidad temporal de este algoritmo de búsqueda es:

$${\mathcal{O}(\log_2 n)}$$
